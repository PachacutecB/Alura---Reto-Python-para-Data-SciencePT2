#📌 Extracción

```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-cursos/challenge2-data-science-LATAM/main/TelecomX_Data.json"
```

```python
df = pd.read_json(url)
display(df.head())
```

#🔧 Transformación

Transforma el dataframe `df` extrayendo la data de las columnas anidadas `customer`, `phone`, `internet`, y `account` usando `pd.json_normalize`.

```python
first_row = df.iloc[0]

nested_columns = ['customer', 'phone', 'internet', 'account']

for col in nested_columns:
    cell_value = first_row[col]
    print(f"Column: {col}")
    print(f"Data type: {type(cell_value)}")
    if isinstance(cell_value, dict):
        print(f"Keys: {cell_value.keys()}")
    print("-" * 20)
```

```python
nested_columns = ['customer', 'phone', 'internet', 'account']
normalized_dfs = []

for col in nested_columns:
    normalized_df = pd.json_normalize(df[col])
    normalized_dfs.append(normalized_df)

df = pd.concat([df.drop(columns=nested_columns)] + normalized_dfs, axis=1)
display(df.head())
```

```python
# Identify columns with '.' which are likely from normalization
cols_to_rename = {col: col.split('.')[-1] for col in df.columns if '.' in col}

# Rename the identified columns
df.rename(columns=cols_to_rename, inplace=True)

# Display the first few rows to verify the changes
display(df.head())
```

#📊 Carga y análisis

```python
print("Data types:")
print(df.dtypes)

print("\nMissing values per column:")
print(df.isnull().sum())
```

#📄Informe final

```python
print("Estadisticas:")
display(df.describe())

print("Categorias:")
for col in df.select_dtypes(include='object').columns:
    print(f"\n--- {col} ---")
    display(df[col].value_counts())
```
